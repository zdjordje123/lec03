# Random Temperature Measurements

Generate 11 random temperature measurements (in Celsius), uniformly distributed between -5°C and 25°C.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n_points = 11
t_min, t_max = -5, 25

temperatures = np.random.uniform(t_min, t_max, n_points)
temperatures = np.round(temperatures, 2)

measurements = np.arange(1, n_points + 1)

for m, t in zip(measurements, temperatures):
    print(f"Measurement {m:2d}: {t:6.2f} \u00b0C")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(measurements, temperatures, color='tab:red', s=80, zorder=3)
plt.plot(measurements, temperatures, color='tab:blue', alpha=0.4, linestyle='--', zorder=2)
plt.axhline(0, color='gray', linewidth=0.8, linestyle=':')

plt.title('Random Temperature Measurements')
plt.xlabel('Measurement #')
plt.ylabel('Temperature (\u00b0C)')
plt.xticks(measurements)
plt.ylim(t_min - 2, t_max + 2)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('temperature_measurements.png', dpi=150)
plt.show()
```
