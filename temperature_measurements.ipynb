# Random Temperature Measurements

Generate 11 random temperature measurements (in Celsius), uniformly distributed between -5°C and 25°C, and convert each to Fahrenheit.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n_points = 11
t_min, t_max = -5, 25

celsius = np.random.uniform(t_min, t_max, n_points)
celsius = np.round(celsius, 2)
fahrenheit = np.round(celsius * 9 / 5 + 32, 2)

measurements = np.arange(1, n_points + 1)

for m, c, f in zip(measurements, celsius, fahrenheit):
    print(f"Measurement {m:2d}: {c:6.2f} \u00b0C  |  {f:6.2f} \u00b0F")
```

```python
fig, ax_c = plt.subplots(figsize=(8, 5))

ax_c.scatter(measurements, celsius, color='tab:red', s=80, zorder=3)
ax_c.plot(measurements, celsius, color='tab:blue', alpha=0.4, linestyle='--', zorder=2)
ax_c.axhline(0, color='gray', linewidth=0.8, linestyle=':')

ax_c.set_title('Random Temperature Measurements')
ax_c.set_xlabel('Measurement #')
ax_c.set_ylabel('Temperature (\u00b0C)', color='tab:red')
ax_c.set_xticks(measurements)
ax_c.set_ylim(t_min - 2, t_max + 2)
ax_c.tick_params(axis='y', labelcolor='tab:red')
ax_c.grid(True, alpha=0.3)

ax_f = ax_c.twinx()
f_min, f_max = t_min * 9 / 5 + 32, t_max * 9 / 5 + 32
ax_f.set_ylim(f_min - 2 * 9 / 5, f_max + 2 * 9 / 5)
ax_f.set_ylabel('Temperature (\u00b0F)', color='tab:green')
ax_f.tick_params(axis='y', labelcolor='tab:green')

fig.tight_layout()
fig.savefig('temperature_measurements.png', dpi=150)
plt.show()
```
